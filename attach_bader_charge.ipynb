### An example of how to attach bader charge to a structural dataset.

```python
# Load external modules
from ase.io import read, write
import os
import numpy as np
import pandas as pd
# Load internal modules
from extract_data.utils import attach_bader_charges

# To use attach_bader_charges you need to provide a set of ZVAl, which you can find from VASP OUTCAR or POTCAR
# Usinf VASP 6.4.0 we found the following zval
zval = {'Na':7,'Fe':14,'O':6,'P':5,'Mn':13,'Co':9,'Ni':16,'Si':4,'S':6}

# Load the data 
atom_without_charge = read('Relax_examples/Na2FeMnNiSiO4/OUTCAR',index=-1)

# Attach the bader charges
acf_path = 'Relax_examples/Na2FeMnNiSiO4/ACF.dat'
atom_with_charge = attach_bader_charges(atom_without_charge,acf_path,zval=zval)

# Extract the charges
bader_charges = atom_with_charge.get_initial_charges()
print({atom.symbol:charge for atom,charge in zip(atom_with_charge,bader_charges)})

# Save the data with the charges
# You need to save the charges in the arrays dictionary of the Atoms object
atom_with_charge.arrays['bader_charge'] = bader_charges
write('Relax_examples/Na2FeMnNiSiO4/atom_with_charge.xyz',atom_with_charge)
```
